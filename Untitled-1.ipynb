```python
def lineups_to_table(lineups, match_id=None):
    rows = []
    for side in ["home", "away"]:
        for entry in lineups[side]["players"]:
            stats = entry.get("statistics", {})
            if not stats:
                continue
            rows.append({
                "match_id": match_id,
                "player": entry["player"]["name"],
                "player_id": entry["player"]["id"],
                "team_id": entry.get("teamId"),
                "side": side,
                "position": entry.get("position"),
                "minutes": stats.get("minutesPlayed", 0),
                "rating": stats.get("rating"),
                "shots": stats.get("totalShots", 0),
                "shots_on_target": stats.get("onTargetScoringAttempt", 0),
                "tackles": stats.get("totalTackle", 0),
                "passes": stats.get("totalPass", 0),
            })
    return pd.DataFrame(rows)
    lineups_to_table(lineups, 16421053)
```

```python
team_id = 2814

url = f"https://www.sofascore.com/api/v1/team/{team_id}/events/last/0"
team_events = requests.get(url, impersonate="chrome").json()

print(len(team_events["events"]))

for e in team_events["events"][-3:]:
    print(e["id"], e["homeTeam"]["name"], "v", e["awayTeam"]["name"], e["startTimestamp"])
```

```python
recent = team_events["events"][-10:]

history = []

for i, event in enumerate(recent):
    match_id = event["id"]
    print(f"{i+1}/10 fetching {match_id}...", end=" ")

    try:
        url = f"https://www.sofascore.com/api/v1/event/{match_id}/lineups"
        response = requests.get(url, impersonate="chrome", timeout=20)

        if response.status_code != 200:
            print(f"skipped ({response.status_code})")
            continue

        table = lineups_to_table(response.json(), match_id)
        history.append(table)
        print(f"{len(table)} players")

    except Exception as exc:
        print(f"failed ({type(exc).__name__})")

    time.sleep(1.5)

all_players = pd.concat(history, ignore_index=True)
print(all_players.shape)
```

```python
LINE = 2

played = espanyol[espanyol["minutes"] > 0].copy()
played["hit"] = played["shots"] >= LINE

hit_rates = played.groupby("player").agg(
    matches=("match_id", "nunique"),
    avg_minutes=("minutes", "mean"),
    avg_shots=("shots", "mean"),
    hits=("hit", "sum"),
).reset_index()

hit_rates["hit_rate_%"] = (hit_rates["hits"] / hit_rates["matches"] * 100).round(0)
hit_rates["avg_shots"] = hit_rates["avg_shots"].round(2)
hit_rates["avg_minutes"] = hit_rates["avg_minutes"].round(0)

hit_rates = hit_rates[hit_rates["matches"] >= 4]
hit_rates.sort_values("hit_rate_%", ascending=False)
```

```python
from pathlib import Path

CACHE = Path("cache")
CACHE.mkdir(exist_ok=True)

BASE = "https://www.sofascore.com/api/v1"


def get_json(path):
    """Fetch an API path, reusing a saved copy if we already have one."""
    name = path.strip("/").replace("/", "_")
    file = CACHE / f"{name}.json"

    if file.exists():
        return json.loads(file.read_text())

    url = f"{BASE}/{path.strip('/')}"
    response = requests.get(url, impersonate="chrome", timeout=20)

    if response.status_code != 200:
        print(f"  {response.status_code} on {path}")
        return None

    data = response.json()
    file.write_text(json.dumps(data))
    time.sleep(1.5)
    return data
```
